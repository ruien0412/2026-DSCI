```python
from os import mkdir

## 將 109 年 ~ 113 年檔案讀入並過濾不必要資料後合併
import pandas as pd
from IPython.display import display
import os

print(os.getcwd())
os.makedirs("data/after_clean", exist_ok=True)

# List to hold individual dataframes
df_list = []

# Loop through years 109 to 113
for year in range(109, 114):
    filename = f'data/before_clean/{year}_大專校院科系別概況.csv'
    try:
        # Read the CSV file
        df = pd.read_csv(filename)

        # Drop specified columns, including '學年度' and '教師數'
        columns_to_drop = ['上學年度畢業生數', '縣市名稱', '體系別', '學年度', '教師數']
        df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

        # Add '年份' column
        df['年份'] = year

        df_list.append(df)
        print(f"Successfully processed {filename}")
    except FileNotFoundError:
        print(f"Warning: File not found for year {year} at {filename}. Skipping.")
    except Exception as e:
        print(f"Error processing {filename}: {e}. Skipping.")

# Concatenate all dataframes into a single one
if df_list:
    combined_df = pd.concat(df_list, ignore_index=True)
    print("\nCombined DataFrame created. Displaying the first 5 rows:")
    display(combined_df.head())
    print("\nCombined DataFrame info:")
    combined_df.info()

    # Save the combined DataFrame to a CSV file
    output_filename = 'data/after_clean/combined_university_data.csv'
    combined_df.to_csv(output_filename, index=False)
    print(f"\nCombined data saved to {output_filename}")
else:
    print("No dataframes were processed successfully.")
```

### 拆分薪資統計資料為博士、碩士、學士三種教育程度的資料框
接下來，我們將 `salary_multiindex_df` 根據教育程度拆分為三個獨立的資料框：碩士、學士和博士。同時，我們將扁平化多層次欄位名稱，以便於後續分析和操作。

```python
import pandas as pd
from pathlib import Path

file_path = 'data/before_clean/薪資統計.csv'
salary_df = pd.read_csv(file_path, encoding='big5')

# 轉長表：把「學士/ xxx學門」這種欄位攤開
long_df = salary_df.melt(
    id_vars=['統計期', '畢業年'],
    var_name='degree_field',
    value_name='薪資'
)

# 拆成「學位」與「學門」
long_df[['學位', '學門']] = long_df['degree_field'].str.split('/', n=1, expand=True)

# 清理空白
long_df['學位'] = long_df['學位'].str.strip()
long_df['學門'] = long_df['學門'].str.strip()

# 不要總計，只留學士/碩士/博士
long_df = long_df[long_df['學位'].isin(['學士', '碩士', '博士'])]

# 如果你不想要「總計」學門，也拿掉
long_df = long_df[long_df['學門'] != '總計']

# 清理薪資欄位
long_df['薪資'] = (
    long_df['薪資']
    .replace(['(D)', '－', '-'], pd.NA)
)

long_df['薪資'] = pd.to_numeric(long_df['薪資'], errors='coerce')

# 重新排序欄位
long_df = long_df[['統計期', '畢業年', '學位', '學門', '薪資']]

display(long_df.head())
```

### 儲存拆分後的資料框為 CSV 檔案
現在，我們將這三個獨立的資料框儲存為各自的 CSV 檔案。

```python
# 儲存為 CSV 檔案
bachelor_salary_df = long_df[long_df['學位'] == '學士']
master_salary_df = long_df[long_df['學位'] == '碩士']
doctorate_salary_df = long_df[long_df['學位'] == '博士']

bachelor_salary_df.to_csv('data/after_clean/薪資統計_學士.csv', index=False, encoding="utf-8-sig")
master_salary_df.to_csv('data/after_clean/薪資統計_碩士.csv', index=False, encoding="utf-8-sig")
doctorate_salary_df.to_csv('data/after_clean/薪資統計_博士.csv', index=False, encoding="utf-8-sig")
```
