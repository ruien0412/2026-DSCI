```python
## 將 109 年 ~ 113 年檔案讀入並過濾不必要資料後合併
import pandas as pd

# List to hold individual dataframes
df_list = []

# Loop through years 109 to 113
for year in range(109, 114):
    filename = f'/content/{year}_大專校院科系別概況.csv'
    try:
        # Read the CSV file
        df = pd.read_csv(filename)

        # Drop specified columns, including '學年度' and '教師數'
        columns_to_drop = ['上學年度畢業生數', '縣市名稱', '體系別', '學年度', '教師數']
        df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

        # Add '年份' column
        df['年份'] = year

        df_list.append(df)
        print(f"Successfully processed {filename}")
    except FileNotFoundError:
        print(f"Warning: File not found for year {year} at {filename}. Skipping.")
    except Exception as e:
        print(f"Error processing {filename}: {e}. Skipping.")

# Concatenate all dataframes into a single one
if df_list:
    combined_df = pd.concat(df_list, ignore_index=True)
    print("\nCombined DataFrame created. Displaying the first 5 rows:")
    display(combined_df.head())
    print("\nCombined DataFrame info:")
    combined_df.info()

    # Save the combined DataFrame to a CSV file
    output_filename = '/content/combined_university_data.csv'
    combined_df.to_csv(output_filename, index=False)
    print(f"\nCombined data saved to {output_filename}")
else:
    print("No dataframes were processed successfully.")
```

### 拆分薪資統計資料為博士、碩士、學士三種教育程度的資料框
接下來，我們將 `salary_multiindex_df` 根據教育程度拆分為三個獨立的資料框：碩士、學士和博士。同時，我們將扁平化多層次欄位名稱，以便於後續分析和操作。

```python
import pandas as pd

# 假設 salary_multiindex_df 已從之前的步驟載入
# 如果此單元格單獨運行，請取消註釋以下行以確保 df 存在：
# file_path = '/content/薪資統計.csv'
# salary_multiindex_df = pd.read_csv(file_path, encoding='big5', header=[2, 3])

# 定義共同的識別欄位
common_identifier_cols = [
    salary_multiindex_df.columns[0],
    salary_multiindex_df.columns[1]
]

def get_degree_dataframe(df, degree_names):
    """
    根據指定的教育程度名稱，從原始 DataFrame 中提取相關欄位並扁平化 MultiIndex。
    degree_names 可以是單個字串或字串列表。
    """
    if isinstance(degree_names, str):
        degree_names = [degree_names]

    selected_cols = []
    # 首先添加共同的識別欄位
    selected_cols.extend(common_identifier_cols)

    # 添加特定教育程度的欄位
    for col_tuple in df.columns:
        if col_tuple[0] in degree_names:
            selected_cols.append(col_tuple)

    degree_df = df[selected_cols].copy()

    # 扁平化 MultiIndex 欄位名稱
    new_columns = []
    for col_tuple in degree_df.columns:
        if col_tuple[0] in [common_identifier_cols[0][0], common_identifier_cols[1][0]]:
            new_columns.append(col_tuple[0]) # 對於共同識別欄位，只保留頂層名稱
        elif pd.isna(col_tuple[1]):
             new_columns.append(col_tuple[0]) # 如果子層名稱為 NaN，則使用頂層名稱
        else:
            # 對於特定教育程度的欄位，將頂層和子層名稱合併，並清理字串
            new_columns.append(f"{col_tuple[0]}_{col_tuple[1].replace(' ', '').replace('/', '_')}")

    degree_df.columns = new_columns

    # 刪除可能因扁平化或資料結構問題而產生的空欄位
    degree_df = degree_df.dropna(axis=1, how='all')

    return degree_df

# 創建碩士學位資料框 (Master's Degree)
# 這包括頂層分類為 '碩士' 和 '碩士(日間)' 的欄位
master_degree_df = get_degree_dataframe(salary_multiindex_df, ['碩士', '碩士(日間)'])
print("\n碩士 DataFrame head:")
display(master_degree_df.head())
print("\n碩士 DataFrame info:")
master_degree_df.info()

# 創建學士學位資料框 (Bachelor's Degree)
# 這包括頂層分類為 '學士' 和 '學士(日間)' 的欄位
bachelor_degree_df = get_degree_dataframe(salary_multiindex_df, ['學士', '學士(日間)'])
print("\n學士 DataFrame head:")
display(bachelor_degree_df.head())
print("\n學士 DataFrame info:")
bachelor_degree_df.info()

# 創建博士學位資料框 (Doctorate Degree)
# 這包括頂層分類為 '博士' 的欄位
doctorate_degree_df = get_degree_dataframe(salary_multiindex_df, '博士')
print("\n博士 DataFrame head:")
display(doctorate_degree_df.head())
print("\n博士 DataFrame info:")
doctorate_degree_df.info()
```
